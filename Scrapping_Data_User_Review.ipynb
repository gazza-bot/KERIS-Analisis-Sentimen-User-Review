Scraping Data User Review - Aplikasi Rumah Pendidikan

```python
import os
import time
import pandas as pd
from google.colab import drive
from google_play_scraper import Sort, reviews

# 1. Mount Google Drive
drive.mount('/content/drive')

# 2. Tentukan Jalur Folder & File
folder_path = '/content/drive/MyDrive/KERIS-Coba/Scrap-Rumah Pendidikan'
file_path = os.path.join(folder_path, 'hasilscrap.csv')

# MEMPERBAIKI ERROR: Buat folder otomatis jika belum ada
os.makedirs(folder_path, exist_ok=True)

# 3. Parameter Scraping
app_id = 'id.belajar.app'
num_reviews = 20000  # Jumlah ulasan yang ingin diambil

# 4. Ambil Ulasan
app_reviews, _ = reviews(
    app_id,
    lang='id',
    country='id',
    sort=Sort.MOST_RELEVANT,
    count=num_reviews
)

# 5. Konversi langsung ke DataFrame Pandas (tidak perlu numpy / pop)
df = pd.DataFrame(app_reviews)

# 6. Simpan DataFrame ke CSV
df.to_csv(file_path, index=False)

print(f"Selesai! Berhasil menyimpan {len(df)} data ulasan ke: {file_path}")
```

```python
df = pd.read_csv(file_path, encoding="utf-8-sig")
df.head()
```

```python
import os
import time
import pandas as pd
from google.colab import drive
from google_play_scraper import Sort, reviews

# 1. Mount Google Drive
drive.mount('/content/drive')

# 2. Tentukan Jalur Folder & File
folder_path = '/content/drive/MyDrive/KERIS-Coba/Scrap-Schoology'
file_path = os.path.join(folder_path, 'hasilscrap.csv')

# MEMPERBAIKI ERROR: Buat folder otomatis jika belum ada
os.makedirs(folder_path, exist_ok=True)

# 3. Parameter Scraping
app_id = 'com.schoology.app'
num_reviews = 20000  # Jumlah ulasan yang ingin diambil

# 4. Ambil Ulasan
app_reviews, _ = reviews(
    app_id,
    lang='id',
    country='id',
    sort=Sort.MOST_RELEVANT,
    count=num_reviews
)

# 5. Konversi langsung ke DataFrame Pandas (tidak perlu numpy / pop)
df = pd.DataFrame(app_reviews)

# 6. Simpan DataFrame ke CSV
df.to_csv(file_path, index=False)

print(f"Selesai! Berhasil menyimpan {len(df)} data ulasan ke: {file_path}")
```
